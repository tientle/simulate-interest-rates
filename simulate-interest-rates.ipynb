Goal: 

Develop a model to set interest rate on a new 3-month CD that your bank wants to roll out. 

In order for you to set the rate, you need to understand the potential range of 3-month interest rate in the future. 

If today’s interest rate is 5%, and changes in interest rate follow a lognormal process with mean zero a standard deviation of 0.625% a day.

Simulate and graph 10 interest rate scenarios path with daily interval.

```python
# Import packages

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter
import math
```

```python
interest_rate = 0.05 # Interest rate is 5%
mean = 0 # Mean is 0
standard_deviation = 0.00625 # Standard deviation is 0.625%
days = 90 # Assume 3 months is about 90 days
simulations = 10 # 10 simulations


# Days + 1 to account for initial day 0
time = np.linspace(0, days, days + 1)

simulated_rates = np.zeros((simulations, days + 1)) # Create empty array for 10 simulations by 90 days (+1 for initial day 0)

# Run simulations
for i in range(simulations):
    normal_increments = np.random.normal(mean, standard_deviation, days)
    log_normal_rates = interest_rate * np.exp(np.cumsum(normal_increments))
    simulated_rates[i, 0] = interest_rate # Day 0 (the initial rate at 5%)
    simulated_rates[i, 1:] = log_normal_rates # simulated rates days 1-91


# Plot the simulation!
plt.figure(figsize=(15,5))

for i in range(simulations):
    plt.plot(time, simulated_rates[i], label=f"Simulation #{i+1}", linewidth=3)

plt.title('Simulated Interest Rates')
plt.xlabel('Time (days)')
plt.ylabel('Interest Rate (%)', rotation=0, labelpad=50)
plt.gca().yaxis.set_major_formatter(PercentFormatter(1))
plt.legend(loc='upper left', bbox_to_anchor=(1.02, 1))

plt.show()
```
